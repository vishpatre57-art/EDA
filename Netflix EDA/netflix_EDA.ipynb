```python
df['director'].fillna('Unknown', inplace=True)
df['cast'].fillna('Unknown', inplace=True)
df['country'].fillna('Unknown', inplace=True)
```

top 10 generes are as belows and top 3 are documentory, standup and comedies dramas and international movies

```python
gen = df['genres'].value_counts().head(10)
plt.barh(gen.index, gen.values)
plt.xlabel('Count',)
plt.ylabel('Genres')
plt.title('Distribution of Genres')
plt.show()
```

```python
con =df['country'].value_counts().head(5)
plt.bar(con.index, con.values)
plt.xlabel('Country',)
plt.ylabel('Count')
plt.title('Distribution of Country')
plt.show()
```

```python
rat = df['rating'].value_counts()
plt.pie(rat.values, labels=rat.index, autopct='%1.1f%%')
plt.title('Distribution of Rating')
plt.show()
```

```python
cast = df[df['cast'] != 'Unknown']['cast'].value_counts().head(10)
plt.barh(cast.index, cast.values)
plt.xlabel('Count',)
plt.ylabel('Cast')
plt.title('Distribution of Cast')
plt.show()
```
