### Check for Duplicates and Nulls

```python
# Check for duplicate rows
duplicate_rows = data.duplicated().sum()
print(f"Number of duplicate rows: {duplicate_rows}")
```

```python
# Check for null values
null_values = data.isnull().sum()
print("\nNull values per column:\n")
print(null_values[null_values > 0])
```

```python
data.describe()
```

```python
data.columns
```

```python
data['rating_count'].fillna(data['rating_count'].mode()[0], inplace=True)
```

```python
After filling the missing values
data.isna().sum()
```

```python
data.sample(5)
```

```python
data['discounted_price']= data['discounted_price'].str.replace('₹', '')
data['discounted_price']= data['discounted_price'].str.replace(',', '')
data['discounted_price']= data['discounted_price'].astype(float)
data['actual_price']= data['actual_price'].str.replace('₹', '')
data['actual_price']= data['actual_price'].str.replace(',', '')
data['actual_price']= data['actual_price'].astype(float)


```

```python
data['discount_percentage'] =data['discount_percentage'].str.replace('%','').astype(float)
data['discount_percentage'] = data['discount_percentage']/100
```

```python
data['rating'].value_counts()
```

```python
data['rating']=data['rating'].str.replace("|",'3.9').astype(float)
```

```python
data['rating_count'] = data['rating_count'].str.replace(',','').astype(float)
```

```python
data.info()
```

```python
data.describe()
```

```python
plt.scatter(data['actual_price'],data['rating'])
plt.xlabel('actual_price')
plt.ylabel('rating')
plt.title('Actual Price vs Rating')
plt.show()
```

```python
plt.hist(data['actual_price'])
plt.xlabel('actual_price')
plt.ylabel('Frequency')
plt.title('Distribution of Actual Price')
plt.show()
#
```

```python
corr_matrix = data.corr(numeric_only=True)
sns.heatmap(corr_matrix, annot=True)
plt.title('Correlation Matrix')
plt.show()
```

```python
sns.countplot(data=data, x='rating')
plt.title('Distribution of Ratings')

plt.show()
```
